```python
import pandas as pd
```

```python
url = 'https://www.contextures.com/xlsampledata01.html' # URL da página que contém a tabela HTML
```

```python
df = pd.DataFrame() 
```

```python
tabela = pd.read_html(url) # O código tabela = pd.read_html(url) é usado para ler tabelas HTML de uma página da web. O argumento url é a URL da página que contém a tabela que você deseja extrair. O método read_html retorna uma lista de DataFrames, onde cada DataFrame corresponde a uma tabela encontrada na página. Você pode acessar a tabela desejada usando o índice da lista, por exemplo, tabela[0] para acessar a primeira tabela encontrada. Certifique-se de que a página da web contenha tabelas HTML para que esse código funcione corretamente.
```

```python
len(tabela) # O código len(tabela) retorna o número de tabelas encontradas na página HTML. No exemplo acima, isso retornará o número 1, indicando que foi encontrada uma tabela na página.
```
