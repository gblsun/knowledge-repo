```python
from sqlalchemy import create_engine # Importa a função 'create_engine' da biblioteca SQLAlchemy, que é usada para criar uma conexão com um banco de dados.
```

```python
engine = create_engine('sqlite:///compras.db') # Cria uma conexão com um banco de dados SQLite chamado 'compras.db' usando a função 'create_engine'.
```

```python
compras.to_sql('compras', con=engine) # Salva o DataFrame 'compras' em uma tabela chamada 'compras' no banco de dados conectado por 'engine'.
```

```python
engine.execute('SELECT * FROM compras').fetchall() # Executa uma consulta SQL para selecionar todos os registros da tabela 'compras' e retorna os resultados como uma lista de tuplas.
```

```python
pd.read_sql_query("SELECT * FROM compras", con=engine) # Lê os dados da tabela 'compras' do banco de dados conectado por 'engine' e retorna um DataFrame com os resultados da consulta SQL.
```
