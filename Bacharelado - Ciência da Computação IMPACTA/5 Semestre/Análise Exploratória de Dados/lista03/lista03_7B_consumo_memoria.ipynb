Durante a análise de performance de um container Docker, foram registrados os seguintes picos de consumo de memória (em MB) durante 7 dias:

**512, 1024, 800, 600, 1024, 1200, 950**

### A)
Determine a mediana do consumo de memória RAM desse container.

### B)
Esboce o box plot.

```python
# Célula 2 - Importação de bibliotecas e definição dos dados
import numpy as np
import matplotlib.pyplot as plt

# Dados originais
consumo = [512, 1024, 800, 600, 1024, 1200, 950]

# Ordenação dos dados
consumo_ordenado = sorted(consumo)
n = len(consumo_ordenado)

print(f"Dados ordenados: {consumo_ordenado}")
print(f"Número de observações (n): {n}")
```

```python
# Célula 3 - Cálculo da mediana e quartis

# Mediana (Q2)
posicao_q2 = (n + 1) / 2
mediana = consumo_ordenado[int(posicao_q2) - 1]

# Quartis
q1 = consumo_ordenado[int((n + 1) / 4) - 1]
q3 = consumo_ordenado[int(3 * (n + 1) / 4) - 1]

print(f"A) Mediana (Q2): {mediana} MB")
print(f"Primeiro Quartil (Q1): {q1} MB")
print(f"Terceiro Quartil (Q3): {q3} MB")
```

```python
# Célula 4 - Cálculo do IIQ e outliers

iiq = q3 - q1
limite_inferior = q1 - 1.5 * iiq
limite_superior = q3 + 1.5 * iiq

print(f"IIQ: {iiq}")
print(f"Limite Inferior: {limite_inferior}")
print(f"Limite Superior: {limite_superior}")

# Outliers
outliers = [x for x in consumo_ordenado if x < limite_inferior or x > limite_superior]
print(f"Outliers encontrados: {outliers}")
```

```python
# Célula 5 - Boxplot

plt.figure(figsize=(8, 6))
plt.boxplot(consumo, vert=True, patch_artist=True)

plt.title("Picos de Consumo de Memória RAM - Container Docker")
plt.ylabel("Consumo (MB)")
plt.xticks([1], ["Container Docker"])
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Anotações
plt.text(1.1, min(consumo), f"Mín: {min(consumo)}")
plt.text(1.1, q1, f"Q1: {q1}")
plt.text(1.1, mediana, f"Mediana: {mediana}")
plt.text(1.1, q3, f"Q3: {q3}")
plt.text(1.1, max(consumo), f"Máx: {max(consumo)}")

plt.show()
```
